```python
with  open("Resource.txt","r",encoding="utf-8") as file:
    noidung = file.read()
print(noidung)
```

```python
with  open("Resource.txt","r",encoding="utf-8") as file:
    noidung = file.read()
print(noidung)
for nd in noidung:
    print(nd)
```

```python
with open("Resource.txt", "r",encoding="utf-8") as file:
    for line in file:
        print(line.strip()) 
```

```python
with open("Resource.txt", "r",encoding="utf-8") as file:
    count = 0
    for line in file:
        count += 1
        print(line.strip())  # Xóa ký tự xuống dòng và khoảng trắng thừa
```
